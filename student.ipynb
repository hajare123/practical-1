```python
employees = {
    101: {"name": "Amit", "department": "IT", "salary": 45000},
    102: {"name": "Priya", "department": "HR", "salary": 38000},
    103: {"name": "Sneha", "department": "IT", "salary": 52000},
    104: {"name": "Raj", "department": "Sales", "salary": 41000},
    105: {"name": "Anjali", "department": "HR", "salary": 47000},
}

for emp_id, data in employees.items():
    print(emp_id, data["name"])
```

```python
for emp_id, data in employees.items():
    if data["salary"] > 40000:
        print(emp_id, data["name"], data["salary"])
```

```python
for emp_id, data in employees.items():
    if data["department"] == "IT":
        print(emp_id, data["name"])
```

```python
max_id = None
max_salary = -1

for emp_id, data in employees.items():
    if data["salary"] > max_salary:
        max_salary = data["salary"]
        max_id = emp_id

print(max_id, employees[max_id]["name"], max_salary)
```

```python
total = 0

for data in employees.values():
    total += data["salary"]

average = total / len(employees)
print(average)
```

```python
products = {
    "P101": {"name": "Laptop", "price": 55000, "stock": 8},
    "P102": {"name": "Mouse", "price": 700, "stock": 25},
    "P103": {"name": "Keyboard", "price": 1200, "stock": 15},
    "P104": {"name": "Monitor", "price": 14000, "stock": 5},
    "P105": {"name": "Headphones", "price": 2500, "stock": 0},
}

for code, data in products.items():
    print(data["name"], data["price"])
```

```python
for code, data in products.items():
    if data["stock"] == 0:
        print(code, data["name"])
```

```python
for code, data in products.items():
    if data["stock"] < 10:
        print(code, data["name"], data["stock"])
```

```python
for code, data in products.items():
    total_value = data["price"] * data["stock"]    
    print(code, data["name"], total_value)
```

```python
for code, data in products.items():    
    if data["price"] > 10000:        
        data["price"] = data["price"] - (data["price"] * 5 / 100)

for code, data in products.items():
    print(code, data["name"], data["price"])
```

```python
text = input("Enter string: ") 
freq = {} 

for ch in text:    
    if ch in freq:        
     freq[ch] += 1    
    else:
        freq[ch] = 1

print(freq)
```

```python
d = {    
    "a": 10,    
    "b": 20,    
    "c": 10,    
    "d": 30,    
}

result = {} 
seen = set()

for key, value in d.items():    
    if value not in seen:        
        result[key] = value
        seen.add(value)
```
