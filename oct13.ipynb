```python
input = [[10,20,30] , [40,50,] , [60,70,80]]
flatten_list = []

for data in input:
    for ele in data:
        flatten_list.append(ele)

print(flatten_list)
```

```python
input = [[10,21,30] , [40,51,] , [60,71,81]]
odd_ele_flatten_list = []

for data in input:
    for ele in data:
        if ele % 2 != 0:
            odd_ele_flatten_list.append(ele)

print(odd_ele_flatten_list)
```
